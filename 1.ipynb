```python
#Common distance measures
    # Euclidean distance:
    # d(x, y) = sqrt( sumi (xi – yi)2 )
    # Squared Euclidean distance:
    # d_sq(x, y) = sumi (xi – yi)2
    # Manhattan distance:
    # d(x, y) = sumi |xi – yi|
    # Minkowski distance:
    # d(x, y) = ( sumi |xi – yi|p )1/p
```

```python
# Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

distance = np.sqrt(np.sum((x - y) ** 2))

print(distance)
```

```python
# Squared Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

squared_distance = np.sum((x - y) ** 2)

print(squared_distance)
```

```python
# Minkowski distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

manhattan_distance = np.sum(np.abs(x - y))

print(manhattan_distance)
```

```python
# Task 2:
import numpy as np

A = np.array([1, 1])
B = np.array([3, 2])
```

```python
# d1(x,y) = |x - y|
def d1(x, y):
    return abs(x - y)

d1(A, B) == d1(B, A)
```

```python
# d2(x, y) = (x - y)^2
def d2(x, y):
    return np.sum((x - y) ** 2)

d2(A, B) == d2(B, A)
```

```python
# d3(x,y) = 0 for all x,y
def d3(x, y):
    return 0

d3(A, B)
```

```python
# d4(x, y) = |x-y| if x >= y, else 2|x-y
def d4(x, y):
    if np.all(x >= y):
        return abs(x - y)
    else:
        return 2 * abs(x - y)

d4(A, B)
```

```python
#  Task 3:
import numpy as np

A = np.array([1.0, 1.0])
B = np.array([1.5, 2.0])
C = np.array([2.0, 1.2])
D = np.array([5.0, 5.0])
E = np.array([6.0, 5.5])
F = np.array([5.5, 6.5])

mu1 = np.array([1.0, 2.8])
mu2 = np.array([6.2, 4.5])

def squ_euc(x, mu):
    return np.sum((x - mu) ** 2)

points = {
    "A": A,
    "B": B,
    "C": C,
    "D": D,
    "E": E,
    "F": F
}

C1 = set()
C2 = set()
J_before = 0

for name, point in points.items():
    d_mu1 = squ_euc(point, mu1)
    d_mu2 = squ_euc(point, mu2)
    
    if d_mu1 < d_mu2:
        C1.add(name)
        J_before += d_mu1
    else:
        C2.add(name)
        J_before += d_mu2
    
print(C1, C2)
```

```python
# One update step: compute new centroids
C1_points = np.array([points[name] for name in C1])
C2_points = np.array([points[name] for name in C2])

new_mu1 = np.mean(C1_points, axis=0)
new_mu2 = np.mean(C2_points, axis=0)


print(new_mu1)
print(new_mu2)
```

```python
# Compute objective after update using the same assignments
J_after = 0

for name in C1:
    J_after += squ_euc(points[name], new_mu1)

for name in C2:
    J_after += squ_euc(points[name], new_mu2)

print()
print("J_before =", round(J_before, 4))
print("J_after =", round(J_after, 4))
```

```python
# Check whether another iteration changes the assignments
new_C1 = set()
new_C2 = set()

for name, point in points.items():
    d_new_mu1 = squ_euc(point, new_mu1)
    d_new_mu2 = squ_euc(point, new_mu2)
    
    if d_new_mu1 < d_new_mu2:
        new_C1.add(name)
    else:
        new_C2.add(name)


print("new C1", new_C1)
print("new C2", new_C2)


print(C1 != new_C1 or C2 != new_C2)
```

```python
# Task 4: K-Means: poor initialisation

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

X = np.array([
    [0.8, 1.2], [1.0, 1.4], [1.2, 0.9], [1.4, 1.6], [0.6, 1.0],
    [1.5, 1.1], [0.9, 1.8], [1.8, 1.3], [1.1, 0.7], [0.7, 1.5],

    [5.8, 1.2], [6.1, 1.5], [6.4, 1.0], [5.6, 1.8], [6.8, 1.4],
    [6.2, 0.7], [5.4, 1.1], [6.7, 1.9], [5.9, 0.6], [6.5, 1.3],

    [3.3, 5.2], [3.7, 5.5], [4.1, 5.9], [4.4, 5.0], [3.5, 6.2],
    [4.6, 5.8], [3.1, 5.8], [4.0, 4.8], [4.8, 5.4], [3.8, 6.4]
])
```

```python
init_centers_poor = np.array([
    [0.8, 1.0],
    [1.4, 1.5],
    [6.0, 1.2]
])

kmeans_poor = KMeans(
    n_clusters=3,
    init=init_centers_poor,
    n_init=1,
    random_state=0
)

labels_poor = kmeans_poor.fit_predict(X)
centers_poor = kmeans_poor.cluster_centers_

print("Final centers with poor initialization:")
print(centers_poor)

print("Inertia with poor initialization:", kmeans_poor.inertia_)
print("Number of iterations:", kmeans_poor.n_iter_)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X[:, 0], X[:, 1], c=labels_poor)

plt.scatter(
    centers_poor[:, 0],
    centers_poor[:, 1],
    marker="X",
    s=200
)

plt.scatter(
    init_centers_poor[:, 0],
    init_centers_poor[:, 1],
    marker="o",
    s=150,
    facecolors="none",
    edgecolors="black"
)

plt.title(f"K-Means with poor initialization, inertia={kmeans_poor.inertia_:.2f}")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
init_centers_better = np.array([
    [1.0, 1.2],   # near left-bottom group
    [6.1, 1.3],   # near right-bottom group
    [4.0, 5.8]    # near top group
])

kmeans_better = KMeans(
    n_clusters=3,
    init=init_centers_better,
    n_init=1,
    random_state=0
)

labels_better = kmeans_better.fit_predict(X)
centers_better = kmeans_better.cluster_centers_

print("Final centers with better initialization:")
print(centers_better)

print("Inertia with better initialization:", kmeans_better.inertia_)
print("Number of iterations:", kmeans_better.n_iter_)
```

```python
init_centers_better = np.array([
    [1.0, 1.2],   # near left-bottom group
    [6.1, 1.3],   # near right-bottom group
    [4.0, 5.8]    # near top group
])

kmeans_better = KMeans(
    n_clusters=3,
    init=init_centers_better,
    n_init=1,
    random_state=0
)

labels_better = kmeans_better.fit_predict(X)
centers_better = kmeans_better.cluster_centers_

print("Final centers with better initialization:")
print(centers_better)

print("Inertia with better initialization:", kmeans_better.inertia_)
print("Number of iterations:", kmeans_better.n_iter_)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X[:, 0], X[:, 1], c=labels_better)

plt.scatter(
    centers_better[:, 0],
    centers_better[:, 1],
    marker="X",
    s=200
)

plt.scatter(
    init_centers_better[:, 0],
    init_centers_better[:, 1],
    marker="o",
    s=150,
    facecolors="none",
    edgecolors="black"
)

plt.title(f"K-Means with better initialization, inertia={kmeans_better.inertia_:.2f}")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(X[:, 0], X[:, 1], c=labels_poor)
axes[0].scatter(centers_poor[:, 0], centers_poor[:, 1], marker="X", s=200)
axes[0].scatter(
    init_centers_poor[:, 0],
    init_centers_poor[:, 1],
    marker="o",
    s=150,
    facecolors="none",
    edgecolors="black"
)
axes[0].set_title(f"Poor init, inertia={kmeans_poor.inertia_:.2f}")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")

axes[1].scatter(X[:, 0], X[:, 1], c=labels_better)
axes[1].scatter(centers_better[:, 0], centers_better[:, 1], marker="X", s=200)
axes[1].scatter(
    init_centers_better[:, 0],
    init_centers_better[:, 1],
    marker="o",
    s=150,
    facecolors="none",
    edgecolors="black"
)
axes[1].set_title(f"Better init, inertia={kmeans_better.inertia_:.2f}")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")

plt.show()
```

```python
# The data set contains three natural groups:
# 1. A left-bottom group around x≈1 and y≈1
# 2. A right-bottom group around x≈6 and y≈1
# 3. A top group around x≈4 and y≈5.5
#
# In the poor initialization, two initial centroids are placed in the left-bottom group,
# and only one centroid is placed in the right-bottom group.
# There is no initial centroid near the top group.
#
# Even though the initialization is poor, K-Means can still move the centroids during
# iterations and may find the three natural groups.
#
# The better initialization places one centroid near each natural group.
# This is more plausible because it matches the visible structure of the data.
#
# The final inertia values can be compared.
# Lower inertia means that points are closer to their assigned centroids.
# In this dataset, both initializations may reach the same final clustering,
# but the better initialization usually reaches it in fewer iterations.
```
