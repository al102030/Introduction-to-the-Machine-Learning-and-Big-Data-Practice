```python
#Common distance measures
    # Euclidean distance:
    # d(x, y) = sqrt( sumi (xi – yi)2 )
    # Squared Euclidean distance:
    # d_sq(x, y) = sumi (xi – yi)2
    # Manhattan distance:
    # d(x, y) = sumi |xi – yi|
    # Minkowski distance:
    # d(x, y) = ( sumi |xi – yi|p )1/p
```

```python
# Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

distance = np.sqrt(np.sum((x - y) ** 2))

print(distance)
```

```python
# Squared Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

squared_distance = np.sum((x - y) ** 2)

print(squared_distance)
```

```python
# Minkowski distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

manhattan_distance = np.sum(np.abs(x - y))

print(manhattan_distance)
```

```python
# Task 2:
import numpy as np

A = np.array([1, 1])
B = np.array([3, 2])
```

```python
# d1(x,y) = |x - y|
def d1(x, y):
    return abs(x - y)

d1(A, B) == d1(B, A)
```

```python
# d2(x, y) = (x - y)^2
def d2(x, y):
    return np.sum((x - y) ** 2)

d2(A, B) == d2(B, A)
```

```python
# d3(x,y) = 0 for all x,y
def d3(x, y):
    return 0

d3(A, B)
```

```python
# d4(x, y) = |x-y| if x >= y, else 2|x-y
def d4(x, y):
    if np.all(x >= y):
        return abs(x - y)
    else:
        return 2 * abs(x - y)

d4(A, B)
```

```python
#  Task 3:
import numpy as np

A = np.array([1.0, 1.0])
B = np.array([1.5, 2.0])
C = np.array([2.0, 1.2])
D = np.array([5.0, 5.0])
E = np.array([6.0, 5.5])
F = np.array([5.5, 6.5])

mu1 = np.array([1.0, 2.8])
mu2 = np.array([6.2, 4.5])

def squ_euc(x, mu):
    return np.sum((x - mu) ** 2)

points = {
    "A": A,
    "B": B,
    "C": C,
    "D": D,
    "E": E,
    "F": F
}

C1 = set()
C2 = set()
J_before = 0

for name, point in points.items():
    d_mu1 = squ_euc(point, mu1)
    d_mu2 = squ_euc(point, mu2)
    
    if d_mu1 < d_mu2:
        C1.add(name)
        J_before += d_mu1
    else:
        C2.add(name)
        J_before += d_mu2
    
print(C1, C2)
```

```python
# One update step: compute new centroids
C1_points = np.array([points[name] for name in C1])
C2_points = np.array([points[name] for name in C2])

new_mu1 = np.mean(C1_points, axis=0)
new_mu2 = np.mean(C2_points, axis=0)


print(new_mu1)
print(new_mu2)
```

```python
# Compute objective after update using the same assignments
J_after = 0

for name in C1:
    J_after += squ_euc(points[name], new_mu1)

for name in C2:
    J_after += squ_euc(points[name], new_mu2)

print()
print("J_before =", round(J_before, 4))
print("J_after =", round(J_after, 4))
```

```python
# Check whether another iteration changes the assignments
new_C1 = set()
new_C2 = set()

for name, point in points.items():
    d_new_mu1 = squ_euc(point, new_mu1)
    d_new_mu2 = squ_euc(point, new_mu2)
    
    if d_new_mu1 < d_new_mu2:
        new_C1.add(name)
    else:
        new_C2.add(name)


print("new C1", new_C1)
print("new C2", new_C2)


print(C1 != new_C1 or C2 != new_C2)
```
