```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])
```

```python
y_pred_1 = 2 * x + 1
y_pred_2 = 1.5 * x + 2
```

```python
print(y_pred_1)
print(y_pred_2)
```

```python
mae_f1 = np.mean(np.abs(y - y_pred_1))
mae_f2 = np.mean(np.abs(y - y_pred_2))

print("MAE for f1 =", mae_f1)
print("MAE for f2 =", mae_f2)

mse_1 = np.mean((y - y_pred_1) ** 2)
mse_2 = np.mean((y - y_pred_2) ** 2)

print("MSE of f1:", mse_1)
print("MSE of f2:", mse_2)
```

```python
x_new = 6

f1_prediction = 2 * x_new + 1
f2_prediction = 1.5 * x_new + 2

print("Prediction of f1 for x=6:", f1_prediction)
print("Prediction of f2 for x=6:", f2_prediction)
```

```python
y = np.array([4, 6, 8, 10])
y_hat = np.array([5, 5, 9, 7])

errors = y_hat - y

mae = np.mean(np.abs(errors))
mse = np.mean(errors ** 2)
mbe = np.mean(errors)

print("Errors:", errors)
print("MAE:", mae)
print("MSE:", mse)
print("MBE:", mbe)
```

## Session 3

```python
#Common distance measures
    # Euclidean distance:
    # d(x, y) = sqrt( sumi (xi – yi)2 )
    # Squared Euclidean distance:
    # d_sq(x, y) = sumi (xi – yi)2
    # Manhattan distance:
    # d(x, y) = sumi |xi – yi|
    # Minkowski distance:
    # d(x, y) = ( sumi |xi – yi|p )1/p
```

```python
# Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

distance = np.sqrt(np.sum((x - y) ** 2))

print(distance)
```

```python
# Squared Euclidean distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

squared_distance = np.sum((x - y) ** 2)

print(squared_distance)
```

```python
# Minkowski distance:
import numpy as np

x = np.array([10, 85])
y = np.array([6, 70])

manhattan_distance = np.sum(np.abs(x - y))

print(manhattan_distance)
```
