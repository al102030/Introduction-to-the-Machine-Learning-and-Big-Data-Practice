```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans


X = np.array([
    [0.8, 1.2], [1.0, 1.4], [1.2, 0.9], [1.4, 1.6], [0.6, 1.0],
    [1.5, 1.1], [0.9, 1.8], [1.8, 1.3], [1.1, 0.7], [0.7, 1.5],

    [5.8, 1.2], [6.1, 1.5], [6.4, 1.0], [5.6, 1.8], [6.8, 1.4],
    [6.2, 0.7], [5.4, 1.1], [6.7, 1.9], [5.9, 0.6], [6.5, 1.3],

    [3.3, 5.2], [3.7, 5.5], [4.1, 5.9], [4.4, 5.0], [3.5, 6.2],
    [4.6, 5.8], [3.1, 5.8], [4.0, 4.8], [4.8, 5.4], [3.8, 6.4]
])
```

```python
init_centers = np.array([[0.8, 1.0], [1.4, 1.5], [6.0, 1.2]])
kmeans = KMeans(n_clusters=3, init=init_centers, n_init=1, random_state=0)
labels = kmeans.fit_predict(X)
centers = kmeans.cluster_centers_
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(centers[:, 0], centers[:, 1], marker='X', s=200)
plt.title(f'K-Means result, inertia={kmeans.inertia_:.2f}')
plt.show()
```
